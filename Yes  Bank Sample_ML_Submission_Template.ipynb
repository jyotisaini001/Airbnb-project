````python
# Appending all models parameters to the corrosponding list

from sklearn.metrics import mean_squared_error
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import r2_score


def score_metrix (model,X_train,X_test,Y_train,Y_test):

  '''
    train the model and gives mae, mse,rmse,r2,adj r2 score of the model

  '''
  #training the model
  model.fit(X_train,Y_train)

  # Training Score
  training  = model.score(X_train,Y_train)
  print("Training score  =", training)

  try:
      # finding the best parameters of the model if any
    print(f"The best parameters found out to be :{model.best_params_} \nwhere model best score is:  {model.best_score_} \n")
  except:
    pass


  #predicting the Test set and evaluting the models

  if model == LinearRegression() or model == Lasso() or model == Ridge():
    Y_pred = model.predict(X_test)

    #finding mean_absolute_error
    MAE  = mean_absolute_error(Y_test**2,Y_pred**2)
    print("MAE :" , MAE)

    #finding mean_squared_error
    MSE  = mean_squared_error(Y_test**2,Y_pred**2)
    print("MSE :" , MSE)

    #finding root mean squared error
    RMSE = np.sqrt(MSE)
    print("RMSE :" ,RMSE)

    #finding the r2 score

    r2 = r2_score(Y_test**2,Y_pred**2)
    print("R2 :" ,r2)
    #finding the adjusted r2 score
    adj_r2=1-(1-r2_score(Y_test**2,Y_pred**2))*((X_test.shape[0]-1)/(X_test.shape[0]-X_test.shape[1]-1))
    print("Adjusted R2 : ",adj_r2,'\n')

  else:
    # for tree base models
    Y_pred = model.predict(X_test)

    #finding mean_absolute_error
    MAE  = mean_absolute_error(Y_test,Y_pred)
    print("MAE :" , MAE)

    #finding mean_squared_error
    MSE  = mean_squared_error(Y_test,Y_pred)
    print("MSE :" , MSE)

    #finding root mean squared error
    RMSE = np.sqrt(MSE)
    print("RMSE :" ,RMSE)

    #finding the r2 score

    r2 = r2_score(Y_test,Y_pred)
    print("R2 :" ,r2)
    #finding the adjusted r2 score
    adj_r2=1-(1-r2_score(Y_test,Y_pred))*((X_test.shape[0]-1)/(X_test.shape[0]-X_test.shape[1]-1))
    print("Adjusted R2 : ",adj_r2,'\n')



  print('*'*80)
  # print the cofficient and intercept of which model have these parameters and else we just pass them
  try :
    print("coefficient \n",model.coef_)
    print('\n')
    print("Intercept  = " ,model.intercept_)
  except:
    pass
  print('\n')
  print('*'*20, 'ploting the graph of Actual and predicted only with 80 observation', '*'*20)

  # ploting the graph of Actual and predicted only with 80 observation for better visualisation which model have these parameters and else we just pass them
  try:
    # ploting the line graph of actual and predicted values
    plt.figure(figsize=(15,7))
    plt.plot((Y_pred)[:80])
    plt.plot((np.array(Y_test)[:80]))
    plt.legend(["Predicted","Actual"])
    plt.show()
  except:
    pass
````

### 1. Which Evaluation metrics did you consider for a positive business impact and why?

For a positive business impact, we primarily consider the following evaluation metrics:

Mean Absolute Error (MAE): MAE represents the average magnitude of errors in a set of predictions, without considering their direction. From a business perspective, a low MAE is highly desirable as it directly indicates the average monetary deviation of our stock price predictions from the actual prices. For instance, an MAE of 6.34 means, on average, our predictions are off by ₹6.34. This is crucial for risk management and setting realistic expectations. Businesses can use MAE to understand the typical error margin in their forecasting, which helps in budgeting, setting stop-loss limits, or determining potential profit margins. A lower MAE leads to more reliable forecasts and, consequently, better financial planning and decision-making.

Root Mean Squared Error (RMSE): RMSE is the square root of the average of the squared differences between prediction and actual observation. It gives a relatively high weight to large errors because the errors are squared before being averaged. In a business context, RMSE helps us understand the magnitude of larger prediction errors. Since stock prices can be highly volatile, large errors can lead to significant financial losses. A low RMSE indicates that the model is not making many large errors, which is critical for minimizing significant financial risks. For example, an RMSE of 9.38 means that the standard deviation of the prediction errors is ₹9.38. This helps in assessing the potential for extreme losses or gains, guiding more cautious investment strategies.

R-squared (R2) and Adjusted R-squared: R2 measures the proportion of the variance in the dependent variable that is predictable from the independent variables. Adjusted R-squared further refines this by adjusting for the number of predictors in the model. Both metrics are crucial for understanding the model's explanatory power and overall reliability. An R2 of 0.99 (or 99%) means that 99% of the variability in the 'Close' price can be explained by our model's features. This indicates a very strong model that captures almost all the relevant patterns in the data.

From a business standpoint, a high R2/Adjusted R2 leads to a high level of confidence in the model's predictions. This confidence enables:

Strategic Investment: Investors can trust the model's insights for making buy/sell decisions, leading to potentially higher returns.
Performance Evaluation: A model with high R2 can be used to set benchmarks and evaluate the effectiveness of trading strategies.
Resource Allocation: Accurate predictions help in allocating resources more effectively, whether it's capital for trading or time for further analysis.
In essence, these metrics collectively ensure that the model is not only accurate on average but also robust against large errors and provides a comprehensive understanding of the factors influencing stock prices. This directly translates into better-informed decisions, reduced risks, and improved financial outcomes for the business.
